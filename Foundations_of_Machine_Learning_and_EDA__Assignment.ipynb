Question 2: Explain overfitting and underfitting in ML. How can you detect and prevent them?
Hint: Discuss bias-variance tradeoff, cross-validation, and regularization techniques.

  -  Definitions and intuition:-

  - Underfitting: Model too simple to capture underlying pattern → high bias. Training error is high and validation error is also high. Example: fitting a straight line to data that is quadratic.

  - Overfitting: Model too complex relative to the data → high variance. Training error is low but validation/test error is high. Example: a high-degree polynomial that wiggles through all training points.

  - Bias–Variance tradeoff

  - Bias: error from wrong assumptions (underfitting).

  - Variance: error from sensitivity to training data (overfitting).
Goal: find model complexity that minimizes total expected error (bias² + variance + irreducible error).

  - How to detect:-

  - Learning curves: plot training and validation error vs. training set size or model complexity.

  - Underfitting: both training & validation errors high and close.

  - Overfitting: training error low, validation error much higher.

  - Cross-validation (k-fold): reliably estimate generalization error; if CV error ≫ training error, likely overfitting.
  
    - How to prevent (practical techniques) :-

  - Simpler model / feature selection: reduce model complexity to lower variance.

  - Regularization: add penalty to weights (L2 ridge, L1 lasso) to shrink coefficients and reduce overfitting.

  - More data: increases ability to generalize (reduces variance).

  - Cross-validation: tune hyperparameters (regularization strength, tree depth, etc.) using CV.

  - Early stopping: stop training (e.g., in neural nets) when validation error starts increasing.

  - Ensemble methods: bagging (reduces variance), boosting (can reduce bias) — often better generalization.

  - Data augmentation: for images/text to synthetically enlarge dataset.

  - Example:

  - If a decision tree with max_depth=20 overfits, set max_depth=5 or apply pruning (reduce complexity) or use RandomForest (bagging) and/or set min_samples_leaf to a larger value.

Question 3: How would you handle missing values in a dataset? Explain at least three methods with examples.
Hint: Consider deletion, mean/median imputation, and predictive modeling.

  - Deletion (listwise or row removal):-

  - When to use: small number of missing rows or missing completely at random (MCAR).

  - Method: drop rows with missing values or drop columns if mostly missing.

  - Example: If only 1% of rows have missing Age, use df.dropna(subset=['Age']).

  - Pros: simple, no bias introduced if MCAR. Cons: loss of data, biased if not MCAR.

  - Simple imputation (mean/median/mode):-

  - When to use: numeric columns with reasonably few missing; median better for skewed data.

  - Method: replace missing numeric values with column mean or median; categorical -> mode.

  - Example (pandas): df['Age'].fillna(df['Age'].median(), inplace=True)

  - Pros: simple, retains dataset size. Cons: underestimates variance and can bias relationships.

  - Predictive (model-based) imputation:-

  - When to use: more accurate than simple imputation if relationships exist.

  - Method: build a model (e.g., regression, k-NN, RandomForest) to predict missing values using other features. Examples:

  - k-NN imputation: fills missing value with average of k nearest neighbors.

  - Multiple imputation (e.g., MICE): creates multiple plausible imputations and pools estimates — better uncertainty handling.

  - Example: Use IterativeImputer (scikit-learn) or fancyimpute or sklearn.impute.KNNImputer.

  - Pros: uses relationships between features; more accurate if model is good. Cons: more complex, risk of leakage if not careful.

  - Bonus methods:

  - Indicator for missingness: create binary flag Age_missing to let model learn any signal in missingness.

  - Domain-specific: fill missing dates with sentinel, or use forward/backward fill for time series.

Question 4: What is an imbalanced dataset? Describe two techniques to handle it (theoretical + practical).
Hint: Discuss SMOTE, Random Under/Oversampling, and class weights in models.
Answer:

Imbalanced dataset

Definition: class distribution skewed (one class much more frequent than others). Common in fraud detection, rare disease diagnosis, churn prediction. Problems: standard classifiers optimize overall accuracy and may ignore minority class → poor recall for minority.

  -  Two techniques:

  - Resampling — Oversampling and Undersampling

    - Random Oversampling: duplicate minority class samples until balanced.

      - Practical: from imblearn.over_sampling import RandomOverSampler then fit/transform.

      - Pros: simple, retains all majority data. Cons: risk of overfitting due to duplicates.

  - Random Undersampling: remove majority class samples to balance.

    - Practical: RandomUnderSampler from imblearn.

    - Pros: faster training. Cons: may discard useful information (loss of data).

  - Synthetic Oversampling — SMOTE (Synthetic Minority Over-sampling Technique)

    - Theory: Create synthetic minority samples by interpolating between minority class neighbors. This reduces exact duplicates and can help decision boundaries generalize.

    - Practical: from imblearn.over_sampling import SMOTE then X_res, y_res = SMOTE().fit_resample(X, y).

    - Pros: reduces overfitting compared to random oversampling. Cons: can create ambiguous samples if classes overlap.

      - Alternative: Class weights / cost-sensitive learning:-

  - Theory: penalize mistakes on minority class more strongly in objective function.

  - Practical: many models accept class_weight='balanced' (e.g., LogisticRegression, RandomForestClassifier) or set weights manually. In XGBoost/LightGBM, specify scale_pos_weight or is_unbalance.

  - Pros: avoids resampling, uses full dataset. Cons: needs careful tuning; sometimes insufficient for extreme imbalance.

  - Evaluation note: use appropriate metrics (precision-recall, F1, ROC AUC, PR AUC) rather than accuracy.

Question 5: Why is feature scaling important in ML? Compare Min-Max scaling and Standardization.
Hint: Explain impact on distance-based algorithms (e.g., KNN, SVM) and gradient descent.


  - Why scaling matters:-

     - Many algorithms use distances (KNN, K-means), assume features on same scale, or rely on gradient-based optimization (logistic regression, neural nets). Unscaled features with large ranges dominate optimization and distance calculations → biased model and slow convergence.

  - Min–Max scaling (Normalization):-

      - Formula: x_scaled = (x - min) / (max - min) → scales to [0,1] (or any [a,b]).

      - Pros: Preserves shape of original distribution, bounded range (useful for algorithms requiring bounded input).

      - Cons: Sensitive to outliers (min or max can be extreme).

      - Best when: features have known bounds or when you need data in a fixed range (e.g., image pixel values).

  - Standardization (Z-score):-

  - Formula: x_scaled = (x - mean) / std → results with mean 0 and standard deviation 1.

  - Pros: Less sensitive to outliers than min-max (but still affected), centers data, works well for algorithms expecting mean 0 (SVM, logistic regression, NN).

  - Cons: not bounded to fixed range.

  - Best when: algorithms assume zero-centered data or when outliers exist.

    - Effect on algorithms:-

  - Distance-based (KNN, K-means): scaling prevents features with larger numeric ranges dominating distance metric.

  - Gradient descent (linear/logistic regression, NN): scaling helps faster and more stable convergence because features have similar gradients.

  - Tree-based models (decision trees, random forests): generally invariant to monotonic scale changes — scaling not strictly necessary.

  - Example: If Age ranges 0–100 and Income 0–1,000,000, without scaling Income dominates Euclidean distance — KNN would effectively ignore Age unless you scale.

Question 6: Compare Label Encoding and One-Hot Encoding. When would you prefer one over the other?
Hint: Consider categorical variables with ordinal vs. nominal relationships.

  - Label Encoding:-

  - What: assign integer codes to categories (e.g., {Red:0, Green:1, Blue:2}).

  - Use-case: ordinal categorical variables where order matters (e.g., size: small<medium<large). Integer mapping preserves ordering for models that can use it.

  - Pros: compact (single column), fast.

  - Cons: introduces spurious ordinal relationships for nominal variables — many algorithms may treat numeric codes as ordered.

  - One-Hot Encoding:-

  - What: create binary dummy columns per category (e.g., is_Red, is_Green, is_Blue).

  - Use-case: nominal categorical variables with no inherent order.

  - Pros: does not impose order, safe for linear models.

  - Cons: increases dimensionality (curse of dimensionality) when categories are many; can cause sparsity. For high-cardinality features consider alternatives (target encoding, embedding).

  - When to prefer which:-

  - Ordinal: prefer Label Encoding or map to numeric order explicitly.

  - Nominal and small cardinality: prefer One-Hot Encoding.

  - High-cardinality nominal: consider target encoding, frequency encoding, or entity embeddings (for neural nets).

  - Example:

  - education_level = ['High School', 'Bachelor', 'Master', 'PhD'] => Label encode (0..3) if order matters.

  - color = ['red','blue','green'] => One-hot encode.

```python
# Code: Google Play category vs rating analysis
# Run in Jupyter/Colab. It reads the googleplaystore.csv from the GitHub repo.
import pandas as pd

# raw URL (replace branch/main if different)
url = "https://raw.githubusercontent.com/MasteriNeuron/datasets/main/googleplaystore.csv"
df = pd.read_csv(url)

# Quick cleaning: convert Rating to numeric, drop rows where Rating is NaN
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
df = df.dropna(subset=['Rating'])

# Group by Category
cat_stats = df.groupby('Category')['Rating'].agg(['count', 'mean', 'median', 'std']).sort_values('mean', ascending=False)
print(cat_stats.head(10))   # top 10 categories by average rating
print("\nLowest average-rating categories:")
print(cat_stats.tail(10))   # bottom 10

# Optional: show boxplots to see distribution
import matplotlib.pyplot as plt
plt.figure(figsize=(12,6))
cat_order = df.groupby('Category')['Rating'].median().sort_values(ascending=False).index
df.boxplot(column='Rating', by='Category', rot=90, figsize=(14,6))
plt.title('Rating distribution by Category')
plt.suptitle('')
plt.show()
```

```python
# Code: Titanic survival analysis by Pclass and Age group
import pandas as pd

url = "https://raw.githubusercontent.com/MasteriNeuron/datasets/main/titanic.csv"
df = pd.read_csv(url)

# Ensure expected columns
# Columns assumed: Survived (0/1), Pclass, Age
print(df[['Survived','Pclass','Age']].head())

# Survival rate by Pclass
pclass_survival = df.groupby('Pclass')['Survived'].agg(['count','sum','mean']).rename(columns={'mean':'survival_rate'})
print("\nSurvival by Pclass:")
print(pclass_survival.sort_index())

# Survival by Age groups: children (<18) vs adults (>=18)
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['age_group'] = df['Age'].apply(lambda x: 'child' if x < 18 else 'adult' if pd.notnull(x) else 'unknown')
age_survival = df[df['age_group'] != 'unknown'].groupby('age_group')['Survived'].agg(['count','mean']).rename(columns={'mean':'survival_rate'})
print("\nSurvival by age group:")
print(age_survival)

# Optional: cross-tab Pclass vs age_group survival rates
ct = pd.crosstab(df['Pclass'], df['age_group'], values=df['Survived'], aggfunc='mean')
print("\nPclass vs age_group survival rates (mean):")
print(ct)
```

  - a) What factors most strongly correlate with employee attrition? Use visualizations to show key drivers (e.g., satisfaction, overtime, salary).
  - b) Are employees with more projects more likely to leave?
  - (Include code / visuals suggestions below.)
    - Answer:
      - Expected insights and interpretation:

  - Typical strong correlates of attrition (depending on dataset): low job satisfaction, high overtime, low salary, few career opportunities, long commute, low work-life balance, low engagement. In many HR datasets, OverTime (yes) is associated with higher attrition rates; low JobSatisfaction correlates with higher attrition.

  - More projects: raw counts may or may not correlate positively with attrition. Sometimes employees with too many projects have higher burnout → increased attrition. In other contexts, more projects can indicate engagement → lower attrition. Use group-by NumProjects and inspect attrition_rate. Visualize as a bar plot of attrition rate vs number of projects to see pattern.

```python
# Code: HR attrition drivers analysis
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/MasteriNeuron/datasets/main/hr_analytics.csv"
df = pd.read_csv(url)

# Assume 'Attrition' is target (Yes/No or 1/0), 'SatisfactionLevel' or 'MonthlyIncome' etc.
# Map Attrition to binary if string
if df['Attrition'].dtype == object:
    df['Attrition_flag'] = df['Attrition'].map({'Yes':1, 'No':0})
else:
    df['Attrition_flag'] = df['Attrition']

# Correlation with numerical features
num_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
corr_with_attrition = df[num_cols].corr()['Attrition_flag'].sort_values(ascending=False)
print("Numeric features correlation with Attrition (top):")
print(corr_with_attrition.head(10))

# Visuals: boxplots / countplots for key drivers
# Example: Satisfaction, OverTime (Yes/No), MonthlyIncome
key_features = ['SatisfactionLevel','OverTime','MonthlyIncome','NumCompaniesWorked','YearsAtCompany','JobSatisfaction','EnvironmentSatisfaction']
for col in key_features:
    if col in df.columns:
        plt.figure(figsize=(6,3))
        if df[col].dtype == 'object' or df[col].nunique() < 6:
            sns.countplot(x=col, hue='Attrition', data=df)
        else:
            sns.boxplot(x='Attrition', y=col, data=df)
        plt.title(f'{col} vs Attrition')
        plt.show()

# b) More projects -> attrition?
if 'NumProjects' in df.columns:
    proj_stats = df.groupby('NumProjects')['Attrition_flag'].agg(['count','mean']).rename(columns={'mean':'attrition_rate'})
    print("\nAttrition rate by number of projects:")
    print(proj_stats)
else:
    print("No 'NumProjects' column. Use 'NumProjects' or equivalent (e.g., 'NumberOfProjects') as in your dataset.")
```
